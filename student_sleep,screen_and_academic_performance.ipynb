```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# 1. Restore the original 'semester_type' column from the zip source
temp_df = pd.read_csv(path)
df['semester_type'] = temp_df['semester_type']

# 2. Impute missing values if any exist in 'semester_type' using the mode
semester_mode = df['semester_type'].mode()[0]
df['semester_type'] = df['semester_type'].fillna(semester_mode)

# 3. Create a dedicated LabelEncoder for 'semester_type'
semester_le = LabelEncoder()

# Fit and transform
df['semester_type_encoded'] = semester_le.fit_transform(df['semester_type'])
print("Encoded unique values:", df['semester_type_encoded'].unique())

# 4. Correctly reverse (inverse transform) back to original labels
df['semester_type_decoded'] = semester_le.inverse_transform(df['semester_type_encoded'])
print("Decoded unique values:", df['semester_type_decoded'].unique())

# Preview the original, encoded, and decoded values side by side
display(df[['semester_type', 'semester_type_encoded', 'semester_type_decoded']].head())
```
